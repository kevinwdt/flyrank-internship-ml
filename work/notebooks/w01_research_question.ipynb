
## 2. The question: decision, action, and cost of a wrong call

My research question is:

**Among pages with enough search impressions, which pages have a lower CTR than expected for pages in a similar average-position range and should be reviewed first?**

The unit of analysis is one pseudonymized webpage.

The decision this project will improve is which pages an SEO specialist, content editor, or content manager should review first. The planned output is a ranked list of possible CTR opportunities.

After reviewing a recommended page, the employee could improve its title or meta description, check whether it matches search intent, improve the content, monitor the page, or decide that no change is necessary.

A false positive would happen if the system recommends a page that does not actually need attention. This could waste time or cause an unnecessary change that lowers clicks.

A false negative would happen if the system misses a real CTR opportunity. The page could continue receiving many impressions without receiving as many clicks as similar pages.

Data may help because a content team can have thousands of pages and cannot review all of them manually. I will begin with a simple rule and only use machine learning later if it creates a more useful and reliable ranking.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*


For this initial check, I define a visible page as one that received at least 500 impressions during the 90-day period and had an average position between 1 and 20.

The starter data contains a large number of visible pages. It also contains pages that received substantial search exposure but recorded no clicks. This suggests that there is a meaningful candidate pool to investigate during the next seven weeks.

These numbers do not prove that the pages have bad titles or need to be changed. They only show that there are enough possible CTR opportunities to make ranking and human review useful. A stronger analysis will compare pages within similar average-position ranges instead of treating all positions equally.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Select pages with enough exposure to make the CTR more meaningful.
visible_pages = df[
    (df["impressions_90d"] >= 500)
    & (df["avg_position"] > 0)
    & (df["avg_position"] <= 20)
].copy()

# Find visible pages that received no clicks in the 90-day period.
zero_click_pages = visible_pages[
    visible_pages["clicks_90d"] == 0
]

zero_click_percentage = (
    len(zero_click_pages) / len(visible_pages) * 100
)

zero_click_impressions = zero_click_pages[
    "impressions_90d"
].sum()

print(f"Total pages in starter dataset: {len(df):,}")

print(
    "Pages with at least 500 impressions and position 1-20:",
    f"{len(visible_pages):,}"
)

print(
    "Visible pages with zero clicks:",
    f"{len(zero_click_pages):,}",
    f"({zero_click_percentage:.1f}% of visible pages)"
)

print(
    "Combined impressions received by those zero-click pages:",
    f"{zero_click_impressions:,.0f}"
)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This project may be able to identify observed patterns between impressions, average position, CTR, clicks, and engagement. It may also help rank pages that deserve human review.

I can describe the results as directional evidence or decision support. I can say that a page received many impressions and had a lower measured CTR than comparable pages.

I cannot claim that:

- a low CTR proves that the title or meta description is bad;
- changing the title will definitely increase clicks;
- the system discovered how Google's ranking algorithm works;
- every page with zero clicks needs to be changed;
- average position and CTR have a causal relationship;
- the results automatically apply to every website or client.

CTR is affected by average position, search intent, search-result features, brand familiarity, query type, and other factors that may not be fully represented in the data. Low-volume pages may also have unstable CTR values. Recommendations should therefore be reviewed by a person before any action is taken.
